```python
numbers = [3, 5, 6, 9, 10, 12, 15, 20, 30]

result = []

for n in numbers:
    if n % 3 == 0 and n % 5 == 0:
        result.append(n)

print("Numbers divisible by both 3 and 5:", result)
```

```python
def second_largest(numbers):
    unique_numbers = list(set(numbers))

    if len(unique_numbers) < 2:
        return "Less than two unique numbers"

    unique_numbers.sort(reverse=True)
    return unique_numbers[1]

numbers = [10, 20, 5, 20, 30, 10, 40]

print("Second largest unique number:", second_largest(numbers))
```

```python
marks = (85, 72, 65, 55, 45, 35)

average = sum(marks) / len(marks)

print("Average:", average)

if average >= 80:
    print("Grade: A")
elif average >= 60:
    print("Grade: B")
elif average >= 40:
    print("Grade: C")
else:
    print("Fail")
```

```python
def duplicate_values(numbers):
    result = []

    for n in numbers:
        if numbers.count(n) > 1 and n not in result:
            result.append(n)

    return tuple(result)

numbers = (1, 2, 3, 2, 4, 5, 3, 6, 2)

print("Duplicate values:", duplicate_values(numbers))
```

```python
def highest_price_product(products):
    if not products:
        return "Dictionary is empty"

    highest_product = max(products, key=products.get)
    return highest_product, products[highest_product]

products = {
    "Laptop": 60000,
    "Phone": 30000,
    "Tablet": 25000,
    "Watch": 10000
}

print("Highest priced product:", highest_price_product(products))
```

```python
students = ["Arun", "Bala", "Chitra", "Divya", "Esha"]
marks = [75, 60, 85, 45, 90]

student_marks = dict(zip(students, marks))

average = sum(marks) / len(marks)

print("Class average:", average)

print("Students above class average:")

for name, mark in student_marks.items():
    if mark > average:
        print(name, mark)
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


start = int(input("Enter starting limit: "))
end = int(input("Enter ending limit: "))

print("Prime numbers:")

for n in range(start, end + 1):
    if is_prime(n):
        print(n)
```

```python
def calculate(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)

    return minimum, maximum, average


numbers = [10, 20, 30, 40, 50]

minimum, maximum, average = calculate(numbers)

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Average:", average)
```

```python
words = ["apple", "banana", "cat", "apple", "dog", "banana"]

unique_words = set(words)

word_lengths = {}

for word in unique_words:
    word_lengths[word] = len(word)

print("Dictionary:", word_lengths)
```

```python
def update_inventory(inventory, item, quantity):
    if item in inventory:
        if inventory[item] >= quantity:
            inventory[item] -= quantity
            print("Order fulfilled")
            print("Updated inventory:", inventory)
        else:
            print("Insufficient stock")
    else:
        print("Item not found")


inventory = {
    "Pen": 20,
    "Book": 10,
    "Pencil": 15
}

item = input("Enter item name: ")
quantity = int(input("Enter quantity: "))

update_inventory(inventory, item, quantity)
```

```python
def calculate_average(records):
    student_marks = {}

    for name, subject, mark in records:
        if name not in student_marks:
            student_marks[name] = []

        student_marks[name].append(mark)

    averages = {}

    for name, marks in student_marks.items():
        averages[name] = sum(marks) / len(marks)

    return averages


records = [
    ("Arun", "Python", 70),
    ("Arun", "SQL", 60),
    ("Bala", "Python", 55),
    ("Bala", "SQL", 65),
    ("Chitra", "Python", 80),
    ("Chitra", "SQL", 75)
]

averages = calculate_average(records)

print("Student averages:")

for name, average in averages.items():
    print(name, average)

print("\nStudents with average 60 or above:")

for name, average in averages.items():
    if average >= 60:
        print(name, average)
```
